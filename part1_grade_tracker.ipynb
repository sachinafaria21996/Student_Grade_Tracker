```python
# Task-1 : Data Parsing and Profile Cleaning

raw_students = [                                               # Given data of students with their names, roll numbers and marks as strings
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

profile_cleaning = []               # For storing the cleaned student profile as a list of dictionaries.


for student in raw_students:                        # Loop through each student in the raw_students list.

    cleaned_name = student["name"].strip()                      #1. Remove extra spaces from the name
    cleaned_name = cleaned_name.title()                         #2. Convert the name to title case (first letter of each word capitalized)
    roll_number = int(student["roll"])                          #3. Convert the roll number to an integer
    
    
    marks = []                                                  #4. Create an empty list to hold our numbers after we clean them up.
    split_marks_list = student["marks_str"].split(",")          #5. Split the string of marks into a list of separate strings, using the comma as a divider.

    for mark in split_marks_list:                               # 6. Loop through the list with split marks
    
        stripped_mark = mark.strip()                            # 7. Remove any extra spaces from each mark string, so we have clean numbers as strings.           
        
        final_integer = int(stripped_mark)                      # 8. Convert that clean text numbers into an integer and store them in final_integer variable.
    
        marks.append(final_integer)                             # 9. Add that integer to our clean_marks list, which will hold all the cleaned up marks as integers.


    # Code for validation of the student names for only alphanumeric characters

    name_letter = cleaned_name.split()                              # 10. Split the cleaned name into individual words

    name_is_valid = all(letter.isalpha() for letter in name_letter) # 11. Check if all characters in the name are alphanumeric (letters only)

    validation_status = "✓ Valid name" if name_is_valid else "✗ Invalid name" # 12. Create a validation status message based on whether the name is valid or not


    # Storing the cleaned data of the student into a dictionary and then appending it to the profile_cleaning list

    cleaned_profile = {
        "name": cleaned_name,
        "roll": roll_number,
        "marks": marks,
        "validation_status": validation_status
    }

    profile_cleaning.append(cleaned_profile)     

    # Printing of the profile card of each student in the given format

    print("============================================")
    print(f"Student : {cleaned_name} {validation_status}")
    print(f"Roll No : {roll_number}")
    print(f"Marks   : {marks}")
    print("============================================ \n")

    #Printing the name name of student roll number 103 in ALL CAPS and lowercase

for student in profile_cleaning:
       if student["roll"] == 103:
           name_upper = student["name"].upper()  # Convert the name to uppercase
           name_lower = student["name"].lower()  # Convert the name to lowercase
           print(f"\nFound student with Roll No 103: {student["name"]}")
           print(f"Name in ALL CAPS: {name_upper}")
           print(f"Name in lowercase: {name_lower}")
           break
```

```python
#TASK2 : Mark Analysis using Loops and Conditionals

# Given Data
student_name = "Ayesha Sharma"
subjects     = ["Math", "Physics", "CS", "English", "Chemistry"]
marks        = [88, 72, 95, 60, 78]

print(f"Grade Card for {student_name} \n")

for i in range(len(subjects)):                     # For Loop to go through the subjects list till the end and store the index in i
 
    subject = subjects[i]                          # Store the name of the subject at index i of subjects list in a variable called subject      
    mark = marks[i]                                # Store the mark at index i of marks list in a variable called mark

    if mark >=90 and mark <=100:                   # Conditionals for grade allocation
        grade = "A+"
    elif mark >=80 and mark <90:
        grade = "A"
    elif mark >=70 and mark <80:
        grade = "B"
    elif mark >=60 and mark <70: 
        grade = "C"
    else :
        grade = "F"

    print(f"Subject : {subject :<15} |  Mark : {mark :<15} (Grade : {grade :<2})")  # Print the subject, mark and grade in the given format

total_marks = sum(marks)                         # Calculate the total marks by summing up all the marks in the marks list
average_marks = round(total_marks/len(marks),2)  # Calculate the average marks by dividing the total marks by the number of subjects (length of marks list) and round it to 2 decimal places
max_marks = max(marks)                           # Getting the maximum marks by finding the highest value in the marks list
min_marks = min(marks)                           # Getting the minimum marks by finding the lowest value in the marks list

highest_subject_index = marks.index(max_marks)   # Getting the index of the highest marks in the marks list, which will correspond to the subject with the highest marks in the subjects list
lowest_subject_index = marks.index(min_marks)    # Getting the index of the lowest marks in the marks list, which will correspond to the subject with the lowest marks in the subjects list

print(f"\nTotal Marks: {total_marks} \nAverage Marks : {average_marks}") 
print(f"Highest Scoring Subject : {subjects[highest_subject_index]} : {max_marks}")  #Print the Highest Scoring subject with its marks
print(f"Lowest Scoring Subject : {subjects[lowest_subject_index]} : {min_marks}")    #Print the Lowest Scoring subject with its marks

# While Loop to enter new subject and marks until the user wants to stop by entering done

initial_subjects_count = len(subjects)  # Store the initial count of subjects to keep track of how many new subjects are added
while True:

    new_subject = input("\nEnter a new subject (or type 'done' to finish): ")
    new_subject = new_subject.strip() 
    
    if new_subject.lower() == 'done':     # Check if the user wants to stop entering new subjects
        break
     
    new_subject_marks = input(f"Enter marks (0 - 100) for {new_subject}: ") # Take input of marks of new subject
    new_subject_marks = new_subject_marks.strip()    # Remove any extra spaces from the input marks

# Validate if the entered marks are a valid integer
    try:                                     # Try block to catch any ValueError that occurs if the input marks cannot be converted to an integer
        new_subject_marks = int(new_subject_marks)
        if 0 <= new_subject_marks <= 100:    # Validate if the marks are between 0 and 100
            subjects.append(new_subject)     # Add the new subject to the subjects list
            marks.append(new_subject_marks)  # Add the new marks to the marks list
            print(f"\nAdded {new_subject} with marks {new_subject_marks}.")

        else:

            print("Do not crash, and do not add invalid entries to the list. \nPlease enter marks between 0 and 100.")
          

    except ValueError:
        
        print("Do not crash, and do not add invalid entries to the list. \nPlease enter a valid integer for marks.")


# Updated Data after adding new subjects and marks
subjects_added = len(subjects) - initial_subjects_count  # Calculate how many new subjects have been added
update_total = sum(marks)            # Update the total marks by summing up all the marks in the updated marks list
update_average = round(update_total/len(marks),2)  # Update the average

print(f"\nSubjects added: {subjects_added} \nUpdated Total Marks: {update_total}")
print(f"\nUpdated Average for {student_name} : {update_average}")
    
```

```python
#TASK 3 - Class Performance Summary

class_data = [                                                          # Given data of students with their names and marks in 5 subjects
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]

passed = 0                                                               # Initialising a variable passed to 0 for counting the number of passed students
failed = 0                                                               # Initialising a variable failed to 0 for counting the number of failed students        
highest_average = 0                                                      # Initialising a variable highest_average to 0 for keeping track of the highest average marks in the class
class_average = 0                                                        # Initialising a variable class_average to 0 for calculating the average marks of the entire class
topper = ""                                                              # Initialising a variable topper to an empty string for storing the name of the class topper 

print(f"{'Name' :<20} | {'Average Marks' :<15} | {'Result Status' :<15}")  # Printing the table heads
print("------------------------------------------")
for student in class_data:                                              # Loop through each student in the class_data list, where each student is a tuple containing the name and marks list
    name = student[0]                                                   # Store the name of the student at index 0 of the student tuple in a variable called name
    marks = student[1]                                                  # Store the marks list of the student at index 1 of the student tuple in a list called marks
    average_marks = round(sum(marks)/len(marks),2)                      # Calculate the average marks of the student by summing up the marks and dividing by the number of subjects (length of marks list) and round it to 2 decimal places
    class_average += (average_marks/len(class_data))                    # Add the average marks of the student to the class_average variable, which will be used to calculate the average marks of the entire class by dividing it by the number of students (length of class_data list)
    if average_marks >=60:
        result_status = "Pass"                                          # Determine the result status of the student based on whether the average marks are greater than or equal to 60
        passed += 1                                                     # Increment the passed variable by 1 if the student has passed                              
    else:
        result_status = "Fail"                                           
        failed += 1                                                     # Increment the failed variable by 1 if the student has failed                           
    
    if average_marks > highest_average:                                 # Check if the average marks of the student are greater than the current highest_average
        highest_average = average_marks                                 # Update the highest_average variable to the average marks of the student if it is greater than the current highest_average
        topper = name                                                   # Update the topper variable to the name of the student if their average marks are greater than the current highest_average
   
 
    print(f"{name :<20} | {average_marks :<15} | {result_status :<15}")

print(f"\nNumber of students passed: {passed} \nNumber of students failed: {failed}")
print(f"\nClass Topper: {topper} with Average Marks : {highest_average}")
print(f"Class Average Marks : {round(class_average,2)}")
```

```python
#TASK 4 String Manipulation Utility

essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

clean_essay = essay.strip()  # Remove leading and trailing whitespace
print(f"\n1. Cleaned Essay: '{clean_essay}'")

#Convert the cleaned essay to title case (first letter of each word capitalized)
clean_essay_converted = clean_essay.title()
print(f"\n2. Essay in Title Case: '{clean_essay_converted}'")

#Count and print number of times python appears in clean essay (case insensitive)
python_count = clean_essay.count("python")
print(f"\n3. Number of times 'python' appears in the essay: {python_count}")

#Replace python with Python 🐍
replaced_essay = clean_essay.replace("python", "Python 🐍")
print(f"\n4. Essay after replacement: '{replaced_essay}'")

#Splitting the essay
essay_split = clean_essay.split(". ")
print(f"\n5. Essay split into sentences and stored as list: {essay_split} ")

#Numbered sentences
print("\n6. Numbered Sentences:")
for i in range(len(essay_split)):         # Loop through the list of sentences in essay_split using the index variable i to keep track of the sentence number.
    sentence = essay_split[i].strip()     # Remove any leading or trailing whitespace from the sentence.

    if not sentence.endswith("."):        # Check if the sentence does not already end with a '.' 
        sentence += "."                   # Add a period at the end of the sentence if it doesn't already end with one.

    print(f"   6.{i+1}. {sentence}")      # Print the sentence in the given format.
```
